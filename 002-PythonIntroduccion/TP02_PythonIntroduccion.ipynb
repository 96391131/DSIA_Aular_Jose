## Ejercicio 1 – Variables y operaciones


Crear variables para almacenar:
tu nombre
tu edad
tu ciudad
Luego imprimir un mensaje con el siguiente formato:
Hola, mi nombre es \_\_\_, tengo \_\_\_ años y vivo en \_\_\_\.
A continuación, crear dos variables numéricas a y b, asignarles valores y calcular:
suma
resta
multiplicación
división
Mostrar los resultados utilizando print\(\)\.


```
#Creación de variables
tu_nombre = "Jose Aular"
tu_edad = 39
tu_ciudad = "Ciudad Autonoma de Buenos Aires"

#Impresión de mensaje por pantalla
print(f'Hola mi nombre es {tu_nombre}, tengo {tu_edad} años y vivo en {tu_ciudad}.')

#Creación de variables numéricas
a = 15
b = 20

#Calculos
suma = a + b
resta = a - b
multiplicacion = a * b
division = a / b

#Muestra de resultados por pantalla
print(f'El resultado de la suma de {a} y {b} es: {suma}')
print(f'El resultado de la resta de {a} y {b} es: {resta}')
print(f'El resultado de la multiplicación de {a} y {b} es: {multiplicacion}')
print(f'El resultado de la división de {a} y {b} es: {division}')
```

## Ejercicio 2 – Condicionales

Crear un programa que:
Pida al usuario que ingrese su edad usando input\(\)
Determine si la persona es mayor o menor de edad
Mostrar uno de los siguientes mensajes:
Eres mayor de edad
o
Eres menor de edad

```
#Petición al usuario del ingreso de su edad
edad = int(input('Ingrese su edad:'))

#Determinación de si la persona es mayor o menor de edad, e impresión del mensaje correspondiente
if edad >= 18:
    print('Eres mayor de edad')
else:
    print('Eres menor de edad')
```

## Ejercicio 3 – Listas y bucles

Crear una lista con 5 números\. Utilizando un bucle for, imprimir cada número de la lista\.
Luego:
Calcular la suma total de los números
Calcular el promedio
Mostrar ambos resultados por pantalla\.

```
#Creación de la lista
numeros=[10, 20, 30, 40, 50]

#Impresión cada número de la lista
print("Lista de números:")
for num in numeros:
    print(num)

#Suma total de los números
suma_total = sum(numeros)

#Promedio de los números
promedio = suma_total / len(numeros)

#Impresión de resultados por pantalla
print(f"Suma total: {suma_total}")
print(f"Promedio: {promedio}")
```
