## Ejercicio 2 – Dataset simple con Pandas

**Crear un DataFrame de Pandas con la siguiente información:**

| estudiante | nota |
| :--- | :---: |
| Ana | 8 |
| Juan | 6 |
| Carla | 9 |
| Pedro | 5 |
| Lucia | 7 |

**Luego realizar las siguientes operaciones:**
1. Mostrar el DataFrame
2. Calcular el promedio de las notas
3. Mostrar la nota máxima
4. Mostrar la nota mínima

````
# Creación  del DataFrame
data = {
    "estudiante": ["Ana", "Juan", "Carla", "Pedro", "Lucia"],
    "nota": [8, 6, 9, 5, 7],
}

df = pd.DataFrame(data)

# 1. Mostrar el DataFrame
print("--- DataFrame ---")
print(df)
print("\n")

# 2. Calcular el promedio de las notas
promedio = df["nota"].mean()
print(f"Promedio de notas: {promedio}")

# 3. Mostrar la nota máxima
nota_maxima = df["nota"].max()
print(f"Nota máxima: {nota_maxima}")

# 4. Mostrar la nota mínima
nota_minima = df["nota"].min()
print(f"Nota mínima: {nota_minima}")
````
